```python
from matplotlib.pyplot import figure, show, tight_layout
from numpy import array, linspace, meshgrid
```

```python
with open('example/output/data.dat', 'r') as file:
    lines = file.readlines()

data = array([float(line) for line in lines]).reshape((3996, 191)).T

with open('example/output/datbis.dat', 'r') as file:
    lines = file.readlines()

databis = array([float(line) for line in lines]).reshape((3996, 191)).T
```

```python
x = linspace(10, 200, 191)
y = linspace(100, 4095, 3996)
XX, YY = meshgrid(y[::-1], x[::-1])
```

```python
fig = figure(figsize=(10, 10))

ax1 = fig.add_subplot(121, projection="3d", proj_type="ortho")
ax2 = fig.add_subplot(122, projection="3d", proj_type="ortho")
ax1.set_title("Original bijection", fontsize=36)
ax2.set_title("Corrected bijection", fontsize=36)
ax1.sharex(ax2)
ax1.sharey(ax2)
ax1.sharez(ax2)
ax1.shareview(ax2)
ax1.set_zlim(-0.5, 0.5)
ax2.set_zlim(-0.5, 0.5)
ax1.set_xlabel("Relative $B_1$ (%)")
ax2.set_xlabel("Relative $B_1$ (%)")
ax1.set_ylabel("$qT_1$ (ms)")
ax2.set_ylabel("$qT_1$ (ms)")
ax1.set_zlabel("$T_1$-w MP2RAGE UNI signal ([-0.5, 0.5])")
ax2.set_zlabel("$T_1$-w MP2RAGE UNI signal ([-0.5, 0.5])")

ax1.view_init(15, -170)
ax2.view_init(15, -170)

# ax = fig.add_subplot(1, 1, 1, projection='3d')
# ax2 = fig.add_subplot(1, 2, 1, projection='3d')
# ax.set_proj_type('ortho')
# # ax.view_init(elev=0, azim=135, roll=0)
# ax2.set_proj_type('ortho')
# # ax2.view_init(elev=0, azim=135, roll=0)
# # ax.view_init(elev=45, azim=135, roll=0)

# Plot a basic wireframe.
ax1.plot_wireframe(XX, YY, data, antialiased=True, alpha=.5, rstride=50, cstride=100)
ax2.plot_wireframe(XX, YY, databis, antialiased=True, alpha=.5, rstride=50, cstride=100)

tight_layout(pad=1)
show()
```
